```python
import pandas as pd

df = pd.read_csv("manufacture cleaned set.csv")

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)
display(df.head())
```

```python
# Dataset dimensions
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# Column names
print("\nALL COLUMN NAMES:")
for i, col in enumerate(df.columns, 1):
    print(i, col)

# Data types
print("\nCOLUMN DATA TYPES:")
print(df.dtypes)

# Dataset preview
print("\nFIRST 5 ROWS:")
display(df.head())

# Dataset information
print("\nDATASET INFORMATION:")
df.info()|
```

```python
df.info()
```

```python
# Missing values count
missing_count = df.isnull().sum()

# Missing percentage
missing_percentage = (missing_count / len(df)) * 100

# Create report
missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": missing_count.values,
    "Missing_Percentage": missing_percentage.values
})

display(missing_report)

print("Total Missing Values:", missing_count.sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Total Records:", len(df))
print("Duplicate Records:", duplicate_count)
print("Unique Records:", len(df) - duplicate_count)
```

```python
# Statistical Analysis

numeric_df = df.select_dtypes(include="number")

statistical_summary = numeric_df.describe().T

display(statistical_summary)

# Additional metrics
additional_stats = pd.DataFrame({
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Standard_Deviation": numeric_df.std(),
    "Variance": numeric_df.var()
})

display(additional_stats)
```

```python
# Outlier Detection using IQR Method

numeric_df = df.select_dtypes(include="number")

outlier_results = []

for col in numeric_df.columns:

    Q1 = numeric_df[col].quantile(0.25)
    Q3 = numeric_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (
        (numeric_df[col] < lower_limit) |
        (numeric_df[col] > upper_limit)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Outlier_Count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)
```

```python
# Categorical Column Distribution

categorical_df = df.select_dtypes(
    include=["object", "category", "bool"]
)

print("Total Categorical Columns:", len(categorical_df.columns))

for col in categorical_df.columns:
    print("\nColumn:", col)
    print(categorical_df[col].value_counts().head(10))
```

```python
import os

# Create reports folder
os.makedirs("reports", exist_ok=True)

# 1. Dataset overview
overview = pd.DataFrame({
    "Metric": ["Total Rows", "Total Columns",
               "Missing Values", "Duplicate Rows"],
    "Value": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

overview.to_csv("reports/dataset_overview.csv", index=False)

# 2. Missing values report
missing_report.to_csv(
    "reports/missing_values.csv", index=False
)

# 3. Statistical summary
statistical_summary.to_csv(
    "reports/statistical_summary.csv"
)

# 4. Outlier report
outlier_report.to_csv(
    "reports/outlier_report.csv", index=False
)

# 5. Categorical distribution
for col in categorical_df.columns:
    categorical_df[col].value_counts(
        dropna=False
    ).to_csv(f"reports/{col}_distribution.csv")

print("All EDA reports saved successfully!")
```
